```python
outlet_columns = [
    'Outlet Size',
    'Outlet Location Type',
    'Outlet Establishment Year'
]

for col in outlet_columns:
    if col in df.columns:
        sales = df.groupby(col)['Sales'].sum().sort_values(ascending=False)

        plt.figure(figsize=(9, 5))
        sns.barplot(x=sales.index.astype(str), y=sales.values)

        plt.title(f"Total Sales by {col}")
        plt.xlabel(col)
        plt.ylabel("Total Sales")
        plt.xticks(rotation=30, ha='right')
        plt.tight_layout()
        plt.show()
```

```python
#ask 3.4: Compare better-performing and lower-performing products and outlets
product_sales = df.groupby('Item Type')['Sales'].sum()

top_products = product_sales.nlargest(5)
bottom_products = product_sales.nsmallest(5)

print("Top 5 Product Categories by Sales:")
print(top_products)

print("\nBottom 5 Product Categories by Sales:")
print(bottom_products)
```

```python
plt.figure(figsize=(12, 6))

sns.barplot(x=top_products.index, y=top_products.values)

plt.title("Top 5 Best-Performing Product Categories")
plt.xlabel("Product Category")
plt.ylabel("Total Sales")
plt.xticks(rotation=30, ha='right')
plt.tight_layout()
plt.show()
```

```python
outlet_sales = df.groupby('Outlet Type')['Sales'].sum()

top_outlets = outlet_sales.nlargest(5)
bottom_outlets = outlet_sales.nsmallest(5)

print("Best-Performing Outlet Types:")
print(top_outlets)

print("\nLower-Performing Outlet Types:")
print(bottom_outlets)
```

```python
plt.figure(figsize=(10, 6))

sns.barplot(x=outlet_sales.sort_values(ascending=False).index,
            y=outlet_sales.sort_values(ascending=False).values)

plt.title("Outlet Performance Comparison")
plt.xlabel("Outlet Type")
plt.ylabel("Total Sales")
plt.xticks(rotation=30, ha='right')
plt.tight_layout()
plt.show()
```

```python
#Help Management Make Better Business Decisions
#Task 4.1: Identify the products that are contributing more to sales
product_sales = df.groupby('Item Type')['Sales'].sum().sort_values(ascending=False)

plt.figure(figsize=(12, 6))
sns.barplot(x=product_sales.index, y=product_sales.values)

plt.title("Product Categories Contributing Most to Sales")
plt.xlabel("Product Category")
plt.ylabel("Total Sales")
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

```python
product_contribution = (
    df.groupby('Item Type')['Sales']
    .sum()
    .sort_values(ascending=False)
    .reset_index()
)

product_contribution['Sales Contribution (%)'] = (
    product_contribution['Sales'] / df['Sales'].sum()
) * 100

print(product_contribution)
```

```python
#Task 4.2: Identify the outlets and locations that are performing better
outlet_type_sales = df.groupby('Outlet Type')['Sales'].sum().sort_values(ascending=False)

plt.figure(figsize=(10, 6))
sns.barplot(x=outlet_type_sales.index, y=outlet_type_sales.values)

plt.title("Sales Performance by Outlet Type")
plt.xlabel("Outlet Type")
plt.ylabel("Total Sales")
plt.xticks(rotation=30, ha='right')
plt.tight_layout()
plt.show()
```

```python
location_sales = df.groupby('Outlet Location Type')['Sales'].sum().sort_values(ascending=False)

plt.figure(figsize=(9, 5))
sns.barplot(x=location_sales.index, y=location_sales.values)

plt.title("Sales Performance by Outlet Location")
plt.xlabel("Outlet Location")
plt.ylabel("Total Sales")
plt.tight_layout()
plt.show()
```

```python
size_sales = df.groupby('Outlet Size')['Sales'].sum().sort_values(ascending=False)

plt.figure(figsize=(9, 5))
sns.barplot(x=size_sales.index, y=size_sales.values)

plt.title("Sales Performance by Outlet Size")
plt.xlabel("Outlet Size")
plt.ylabel("Total Sales")
plt.tight_layout()
plt.show()
```

```python
outlet_performance = df.groupby('Outlet Identifier').agg(
    Total_Sales=('Sales', 'sum'),
    Outlet_Type=('Outlet Type', 'first'),
    Location=('Outlet Location Type', 'first'),
    Outlet_Size=('Outlet Size', 'first')
)

print(outlet_performance.sort_values('Total_Sales', ascending=False))
```

```python
average_outlet_sales = outlet_performance.groupby('Outlet_Type')['Total_Sales'].mean().sort_values(ascending=False)

print("Average Sales per Outlet by Outlet Type:")
print(average_outlet_sales)
```

```python
#Task 4.3: Identify products, outlets, and areas where performance can be improved
product_sales = df.groupby('Item Type')['Sales'].sum().sort_values()

bottom_products = product_sales.head(5)

plt.figure(figsize=(10, 6))
sns.barplot(x=bottom_products.values, y=bottom_products.index)

plt.title("Five Lowest-Selling Product Categories")
plt.xlabel("Total Sales")
plt.ylabel("Product Category")
plt.tight_layout()
plt.show()

print(bottom_products)
```

```python
bottom_outlets = outlet_performance.nsmallest(10, 'Total_Sales')

print("10 Lowest-Selling Outlets:")
print(bottom_outlets)

plt.figure(figsize=(12, 6))

sns.barplot(
    x=bottom_outlets.index.astype(str),
    y=bottom_outlets['Total_Sales']
)

plt.title("10 Lowest-Selling Outlets")
plt.xlabel("Outlet Identifier")
plt.ylabel("Total Sales")
plt.tight_layout()
plt.show()
```

```python
location_performance = df.groupby('Outlet Location Type').agg(
    Total_Sales=('Sales', 'sum'),
    Average_Sales=('Sales', 'mean'),
    Transaction_Count=('Sales', 'count')
).sort_values('Total_Sales')

print(location_performance)
```

```python
#Task 4.4: Provide practical suggestions to improve sales and overall business performance
top_products = df.groupby('Item Type')['Sales'].sum().nlargest(5)

print("Top 5 Product Categories:")
print(top_products)

top_outlet_types = df.groupby('Outlet Type')['Sales'].sum().sort_values(ascending=False)

print("Sales by Outlet Type:")
print(top_outlet_types)


fat_sales = df.groupby('Item Fat Content')['Sales'].sum().sort_values(ascending=False)

plt.figure(figsize=(8, 5))
sns.barplot(x=fat_sales.index, y=fat_sales.values)

plt.title("Sales by Item Fat Content")
plt.xlabel("Item Fat Content")
plt.ylabel("Total Sales")
plt.tight_layout()
plt.show()


best_product = product_sales.idxmax()
lowest_product = product_sales.idxmin()

best_outlet_type = outlet_type_sales.idxmax()
best_location = location_sales.idxmax()

print("BUSINESS PERFORMANCE SUMMARY")
print("-" * 40)

print(f"Highest-selling product category: {best_product}")
print(f"Lowest-selling product category: {lowest_product}")
print(f"Best-performing outlet type by total sales: {best_outlet_type}")
print(f"Best-performing location by total sales: {best_location}")

print("\nRecommended Actions:")
print("1. Maintain adequate stock of high-selling product categories.")
print("2. Investigate low-selling categories before changing inventory.")
print("3. Study the practices of high-performing outlets.")
print("4. Review product availability and customer demand at low-performing outlets.")
print("5. Evaluate promotions using changes in sales and profit margins.")
```
